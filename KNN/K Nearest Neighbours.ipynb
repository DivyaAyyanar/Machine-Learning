# KNN - K Nearest Neighbours

## 1. Import Necessary Libraries

```python
import pandas as pd
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler
```

## 2. Data Collection

```python
load_wine()
```

```python
wine_data = load_wine()
wine_data_df = pd.DataFrame(wine_data.data, columns = wine_data.feature_names)
wine_data_df["target"] = wine_data.target
wine_data_df
```

## 3. Data Understanding

```python
wine_data_df.isna().sum()
```

```python
wine_data_df.dtypes
```

## 4. Data Preparation

##### Data cleaning and data transformation are not needed because, no need to drop any columns, no need to change datatypes and the input featues are numerical

## 5. Model Building

```python
X = wine_data_df.drop("target", axis = 1)
y = wine_data_df["target"]
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.20,random_state = 123, shuffle = True, stratify = y)
```

## 6. Model Training // 7. Model Testing // 8. Model Evaluation

### When K=3

#### Training Data

```python
knn_model = KNeighborsClassifier(n_neighbors = 3)
knn_model.fit(X_train, y_train)
y_pred_train = knn_model.predict(X_train)
print("Training Accuracy:", round(accuracy_score(y_train,y_pred_train),4))
```

#### Test Data

```python
knn_model_test = KNeighborsClassifier(n_neighbors = 3)
knn_model_test.fit(X_test, y_test)
y_pred_test = knn_model_test.predict(X_test)
print("Test Accuracy:", round(accuracy_score(y_test, y_pred_test),4))
```

### When K=5

#### Training Data

```python
knn_model = KNeighborsClassifier(n_neighbors = 5)
knn_model.fit(X_train, y_train)
y_pred_train = knn_model.predict(X_train)
print("Training Accuracy:", round(accuracy_score(y_train,y_pred_train),4))
```

#### Test Data

```python
knn_model_test = KNeighborsClassifier(n_neighbors = 5)
knn_model_test.fit(X_test, y_test)
y_pred_test = knn_model_test.predict(X_test)
print("Test Accuracy:", round(accuracy_score(y_test, y_pred_test),4))
```

### When K=7
