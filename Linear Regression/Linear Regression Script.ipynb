### Data Preparation:

```python
X = car_dataset[["HP","VOL","SP","WT"]]
X
```

```python
y=car_dataset["MPG"]
y
```

### Model Building:

```python
linear_model=LinearRegression()
```

### Model Training:

```python
linear_model.fit(X,y)
```

```python
linear_model.intercept_
```

```python
linear_model.coef_
```

### Model Testing:

```python
y_pred = linear_model.predict(X)
y_pred 
```

### Model Evaluation:

```python
error = y-y_pred
error
```

### Homoscadasticity:

```python
sns.scatterplot(data=car_dataset,x="HP",y=error)
```

```python
sns.scatterplot(data=car_dataset,x="VOL",y=error)
```

```python
sns.scatterplot(data=car_dataset,x="SP",y=error)
```

```python
sns.scatterplot(data=car_dataset,x="WT",y=error)
```
