# SVM Support Vector Machine

## 1. Import necessary Libraries

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix
```

## 2. Data Collection

```python
breast_cancer_data = load_breast_cancer()
breast_cancer_data_df = pd.DataFrame(breast_cancer_data.data, columns = breast_cancer_data.feature_names)
breast_cancer_data_df["Target"] = breast_cancer_data.target
breast_cancer_data_df
```

## 3.Data Understanding

```python
breast_cancer_data_df.isna().sum()
```

```python
breast_cancer_data_df.dtypes
```

## 4. Data Preparation

##### Data cleaning and data transformation are not needed because, no need to drop any columns, no need to change datatypes and the input featues are numerical

## 5. Model Building

```python
X = breast_cancer_data_df.drop("Target", axis = 1)
y = breast_cancer_data_df["Target"]
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.20,random_state = 123, shuffle = True, stratify = y)
```

```python
X_train.shape,y_train.shape
```

```python
X_test.shape,y_test.shape
```

## 6. Model Training // 7. Model Testing // 8. Model Evaluation

### Linear 

#### Training Data

```python
svc_model = SVC(kernel='linear')
svc_model.fit(X_train, y_train)
y_pred_train = svc_model.predict(X_train)
print("Training Accuracy:", round(accuracy_score(y_train,y_pred_train),4))
print("Training Confusion Matrix:\n", confusion_matrix(y_train,y_pred_train))
```

#### Test Data

```python
svc_model = SVC(kernel='linear')
svc_model.fit(X_test, y_test)
y_pred_test = svc_model.predict(X_test)
print("Test Accuracy:", round(accuracy_score(y_test,y_pred_test),4))
print("Test Confusion Matrix:\n", confusion_matrix(y_test,y_pred_test))
```

#### Training Data

```python
svc_model = SVC(kernel='poly')
svc_model.fit(X_train, y_train)
y_pred_tra1in = svc_model.predict(X_train)
print("Training Accuracy:", round(accuracy_score(y_train,y_pred_train),4))
print("Training Confusion Matrix:\n", confusion_matrix(y_train,y_pred_train))
```

#### Test Data

```python
svc_model = SVC(kernel='poly')
svc_model.fit(X_test, y_test)
y_pred_test = svc_model.predict(X_test)
print("Test Accuracy:", round(accuracy_score(y_test,y_pred_test),4))
print("Test Confusion Matrix:\n", confusion_matrix(y_test,y_pred_test))
```

#### Training Data

```python
svc_model = SVC(kernel='rbf')
svc_model.fit(X_train, y_train)
y_pred_train = svc_model.predict(X_train)
print("Training Accuracy:", round(accuracy_score(y_train,y_pred_train),4))
print("Training Confusion Matrix:\n", confusion_matrix(y_train,y_pred_train))
```

```python
svc_model = SVC(kernel='rbf')
svc_model.fit(X_test, y_test)
y_pred_test = svc_model.predict(X_test)
print("Test Accuracy:", round(accuracy_score(y_test,y_pred_test),4))
print("Test Confusion Matrix:\n", confusion_matrix(y_test,y_pred_test))
```
